```python
import pandas as pd
import numpy as np
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import re
import json
import warnings
warnings.filterwarnings('ignore')
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')
```

```python
class FinalYearProjectSelector:
    def __init__(self):
        self.lemmatizer = WordNetLemmatizer()
        self.stop_words = set(stopwords.words('english'))
        self.projects = self.load_projects()
        self.students = self.load_students()
        self.vectorizer = TfidfVectorizer(max_features=1000, stop_words='english')
        self.project_vectors = self.create_project_vectors()

    def preprocess_text(self, text):
        text = text.lower()
        text = re.sub(r'[^a-zA-Z\s]', '', text)
        tokens = nltk.word_tokenize(text)
        tokens = [self.lemmatizer.lemmatize(token) for token in tokens
                    if token not in self.stop_words and len(token) > 2]
        return ' '.join(tokens)

    def load_projects(self):
        data = {
            'project_id': ['P001', 'P002', 'P003', 'P004', 'P005',
                         'P006', 'P007', 'P008', 'P009', 'P010',
                         'P011', 'P012', 'P013', 'P014', 'P015'],
            'title': [
                'AI-Powered Mental Health Chatbot',
                'Predictive Maintenance in Manufacturing',
                'Real-Time Traffic Prediction System',
                'Fake News Detection with NLP',
                'Smart Agriculture IoT Platform',
                'E-Commerce Recommendation Engine',
                'Medical Image Classification with CNN',
                'Customer Churn Prediction',
                'Sentiment Analysis on Social Media',
                'Autonomous Driving Object Detection',
                'Blockchain-Based Supply Chain',
                'Intelligent Tutoring System',
                'Fraud Detection in Banking',
                'Energy Consumption Forecasting',
                'Generative AI Art Generator'
            ],
            'domain': [
                'Healthcare', 'Industrial', 'Smart City',
                'NLP', 'Agriculture', 'E-Commerce',
                'Healthcare', 'Finance', 'Social Media',
                'Automotive', 'Blockchain', 'Education',
                'Finance', 'Energy', 'Generative AI'
            ],
            'tech_stack': [
                'Python, NLP, Flask',
                'Python, Scikit-learn, Pandas',
                'Python, TensorFlow, Kafka',
                'Python, NLTK, Transformers',
                'Python, IoT, Django',
                'Python, Pandas, Surprise',
                'Python, TensorFlow, OpenCV',
                'Python, XGBoost, Pandas',
                'Python, NLTK, Matplotlib',
                'Python, YOLO, OpenCV',
                'Python, Flask, Web3',
                'Python, Django, React',
                'Python, PyTorch, Pandas',
                'Python, Prophet, Pandas',
                'Python, PyTorch, Stable Diffusion'
            ],
            'difficulty': [
                'Intermediate', 'Advanced', 'Intermediate',
                'Beginner', 'Intermediate', 'Intermediate',
                'Advanced', 'Beginner', 'Beginner',
                'Advanced', 'Intermediate', 'Intermediate',
                'Intermediate', 'Intermediate', 'Advanced'
            ],
            'duration': [
                '6 weeks', '8 weeks', '6 weeks',
                '4 weeks', '6 weeks', '6 weeks',
                '8 weeks', '4 weeks', '4 weeks',
                '10 weeks', '6 weeks', '6 weeks',
                '6 weeks', '6 weeks', '8 weeks'
            ],
            'description': [
                'Build an AI chatbot for mental health support with NLP and sentiment analysis.',
                'Predict equipment failures using machine learning to reduce downtime.',
                'Develop a system to predict traffic congestion in real-time.',
                'Detect fake news articles using natural language processing techniques.',
                'Create an IoT platform for smart agriculture and crop monitoring.',
                'Build a recommendation engine for personalized product suggestions.',
                'Classify medical images using convolutional neural networks.',
                'Predict customer churn to improve retention strategies.',
                'Analyze sentiment from social media data using NLP.',
                'Implement object detection for autonomous vehicles using YOLO.',
                'Create a blockchain-based supply chain tracking system.',
                'Develop an intelligent tutoring system with personalized learning.',
                'Build a fraud detection system for banking transactions.',
                'Forecast energy consumption using time series analysis.',
                'Create a generative AI art generator using Stable Diffusion.'
            ]
        }
        df = pd.DataFrame(data)
        df['combined_text'] = df['title'] + ' ' + df['domain'] + ' ' + df['tech_stack'] + ' ' + df['description']
        df['clean_text'] = df['combined_text'].apply(self.preprocess_text)
        return df

    def load_students(self):
        """ Load sample student profiles """
        np.random.seed(42)
        data = {
            'student_id': ['S001', 'S002', 'S003', 'S004', 'S005'],
            'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
            'domain_interest': ['AI', 'Web', 'Data Science', 'AI', 'Mobile'],
            'skills': [
                ['Python', 'NLP', 'Machine Learning'],
                ['Python', 'Django', 'React'],
                ['Python', 'Pandas', 'Statistics'],
                ['Python', 'TensorFlow', 'Computer Vision'],
                ['Python', 'Flutter', 'Firebase']
            ],
            'interests': [
                ['Healthcare', 'NLP', 'Chatbot'],
                ['E-Commerce', 'Web Development', 'Blockchain'],
                ['Finance', 'Analytics', 'Prediction'],
                ['Automotive', 'Computer Vision', 'Autonomous'],
                ['Education', 'Mobile Development', 'Tutoring']
            ],
            'experience_level': ['Intermediate', 'Advanced', 'Intermediate', 'Advanced', 'Beginner']
        }
        df = pd.DataFrame(data)
        df['profile_text'] = df['domain_interest'] + ' ' + df['skills'].apply(lambda x: ' '.join(x)) + ' ' + df['interests'].apply(lambda x: ' '.join(x))
        df['clean_profile'] = df['profile_text'].apply(self.preprocess_text)
        return df

    def create_project_vectors(self):
        """ Create TF-IDF vectors for all projects """
        return self.vectorizer.fit_transform(self.projects['clean_text'])

    def recommend_projects(self, student_id, top_n=5):
        """ Recommend projects for a specific student """
        if student_id not in self.students['student_id'].values:
            return "Student not found"

        student = self.students[self.students['student_id'] == student_id].iloc[0]
        student_profile = student['clean_profile']
        student_vector = self.vectorizer.transform([student_profile])

        """ Calculate similarity with all projects """
        similarities = cosine_similarity(student_vector, self.project_vectors)[0]

        """ Get top N projects """
        top_indices = np.argsort(similarities)[::-1][:top_n]

        recommendations = []
        for idx in top_indices:
            project = self.projects.iloc[idx]
            recommendations.append({
                'project_id': project['project_id'],
                'title': project['title'],
                'domain': project['domain'],
                'tech_stack': project['tech_stack'],
                'difficulty': project['difficulty'],
                'duration': project['duration'],
                'description': project['description'],
                'similarity_score': round(similarities[idx], 3)
            })

        return recommendations

    def filter_by_domain(self, recommendations, domain):
        """ Filter recommendations by domain """
        return [rec for rec in recommendations if rec['domain'] == domain]

    def filter_by_difficulty(self, recommendations, level):
        """ Filter recommendations by difficulty level """
        return [rec for rec in recommendations if rec['difficulty'] == level]

    def get_project_details(self, project_id):
        """ Get detailed information about a specific project """
        project = self.projects[self.projects['project_id'] == project_id]
        if len(project) == 0:
            return "Project not found"
        return project.iloc[0].to_dict()

    def generate_report(self, student_id):
        """ Generate a comprehensive report with recommendations """
        recommendations = self.recommend_projects(student_id)
        if isinstance(recommendations, str):
            return recommendations

        student = self.students[self.students['student_id'] == student_id].iloc[0]

        report = f"""
Project Recommendation Report for {student['name']} (ID: {student['student_id']})

Student Profile:
  Domain Interest: {student['domain_interest']}
  Skills: {', '.join(student['skills'])}
  Interests: {', '.join(student['interests'])}
  Experience Level: {student['experience_level']}

Top Recommended Projects:
"""
        for i, rec in enumerate(recommendations):
            report += f"""
  {i+1}. {rec['title']} (ID: {rec['project_id']})
     Domain: {rec['domain']}
     Tech Stack: {rec['tech_stack']}
     Difficulty: {rec['difficulty']}
     Duration: {rec['duration']}
     Similarity Score: {rec['similarity_score']}
     Description: {rec['description']}
"""
        return report
```

```python
# Usage Example
selector = FinalYearProjectSelector()

# Get project recommendations for a student
print("🎓 Final Year Project Recommendations for S001:")
print("="*50)
recommendations = selector.recommend_projects("S001", top_n=5)

for i, rec in enumerate(recommendations, 1):
    print(f"\n{i}. {rec['title']}")
    print(f"   📂 Domain: {rec['domain']}")
    print(f"   🔧 Tech: {rec['tech_stack']}")
    print(f"   ⚡ Difficulty: {rec['difficulty']}")
    print(f"   📊 Match: {rec['similarity_score']*100:.1f}%")

# Generate comprehensive report
print("\n📄 Comprehensive Report:")
print(selector.generate_report("S001"))

# Get recommendations for another student
print("\n🎓 Recommendations for S003 (Data Science Interest):")
print("="*50)
recs = selector.recommend_projects("S003", top_n=3)
for i, rec in enumerate(recs, 1):
    print(f"{i}. {rec['title']} ({rec['domain']}) - {rec['similarity_score']*100:.1f}%")

# Filter recommendations by domain
print("\n🔍 Filtered by Domain (Healthcare):")
print("="*50)
healthcare = selector.filter_by_domain(recommendations, 'Healthcare')
for rec in healthcare:
    print(f"  - {rec['title']}")

# Filter recommendations by difficulty
print("\n⚡ Filtered by Difficulty (Advanced):")
print("="*50)
advanced = selector.filter_by_difficulty(recommendations, 'Advanced')
for rec in advanced:
    print(f"  - {rec['title']}")
```
